## 2) Create InChIKeys based on SMILES structure 

#### 1. **Standardize and Clean SMILES Representations**
- Remove whitespace, erroneous entries (e.g. `'nan'`, `'N/A'`, `'InChI=...'`), and ensure consistent formatting of SMILES strings in both GNPS and MIBiG datasets.
- This prevents RDKit from parsing invalid strings and minimizes noise in later steps.

#### 2. **Generate InChIKeys from SMILES**
- Use RDKit’s `Chem.MolFromSmiles()` and `inchi.MolToInchiKey()` functions to generate standardized **InChIKeys** for all valid SMILES entries.

#### 3. **Extract the Connectivity Layer**
- From each full InChIKey (typically a 27-character string in the form `AAAAAA-BBBBB-CC`), extract the first 14-character block (before the first dash).
- This **connectivity layer** encodes atom connectivity (i.e. molecular graph structure) and is robust to stereochemistry and protonation differences.

#### 4. **Match Structures Between Datasets**
- Remove rows with missing connectivity blocks in either dataset.
- Drop duplicates in MIBiG to ensure a one-to-one mapping from `inchikey_connectivity` to its `accession`.

#### 5. **Merge and Inspect Results**
- The merged dataset contains GNPS spectra enriched with biosynthetic gene cluster (BGC) annotations from MIBiG based on molecular structure similarity.
